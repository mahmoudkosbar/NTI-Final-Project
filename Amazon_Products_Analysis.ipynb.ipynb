##**Data Cleaning**

```python
# Check missing values
missing_values = df.isnull().sum().sort_values(ascending=False)

missing_values
```

```python
# Missing values percentage
missing_percentage = (df.isnull().mean() * 100).sort_values(ascending=False)

missing_percentage
```

```python
# Check number of duplicate rows
duplicate_count = df.duplicated().sum()

print(f"Number of duplicate rows: {duplicate_count}")
```

```python
# delet sustainability_badges
df = df.drop(columns=['sustainability_badges'])
```

```python
# Missing values percentage
missing_percentage = (df.isnull().mean() * 100).sort_values(ascending=False)

missing_percentage
```

```python
# Fill missing string values

columns_to_fill = [
    'buy_box_availability',
    'delivery_details',
    'product_url',
    'image_url',
    'is_sponsored',
    'listed_price',
    'is_best_seller',
    'is_couponed',
    'collected_at',
    'price_on_variant'
]

df[columns_to_fill] = df[columns_to_fill].fillna('Not Available')
```

```python
df.info()
```

```python
# اي شكل البيانات في الاعمدة دي عشان افكر ازاي انضفها
columns_to_inspect = [
    'rating',
    'number_of_reviews',
    'bought_in_last_month',
    'current/discounted_price'
]

for col in columns_to_inspect:
    print("\n" + "="*60)
    print(col)
    print("dtype:", df[col].dtype)
    print("Missing:", df[col].isna().sum())
    print("Examples:")
    print(df[col].dropna().head(10).tolist())
```

```python
# bought_in_last_month
df['bought_in_last_month'] = df['bought_in_last_month'].fillna('Not Available')
df['bought_in_last_month'].isnull().sum()
```

```python
# convert rating from '4.6 out of 5 stars' to 4.6
import re
df['rating'] = (
    df['rating']
    .str.extract(r'(\d+\.?\d*)')[0]
    .astype(float)
)

df["rating"]
```

```python
# convert num of review >> 35,882 to 35882
df['number_of_reviews'] = (
    df['number_of_reviews']
    .str.replace(',', '', regex=False)
    .astype(float)
)
df['number_of_reviews']
```

```python
# current/discounted_price
df['current/discounted_price'] = (
    df['current/discounted_price']
    .str.replace(',', '', regex=False)
)

df['current/discounted_price'] = pd.to_numeric(
    df['current/discounted_price'],
    errors='coerce'
)

df['current/discounted_price']
```

```python
# Handling numeric
numeric_columns = [
    'rating',
    'number_of_reviews',
    'current/discounted_price'
]

for col in numeric_columns:
    median_value = df[col].median()
    df[col] = df[col].fillna(median_value)
    print(f"{col} → Median used: {median_value}")
```

```python
# Missing values percentage
missing_percentage = (df.isnull().mean() * 100).sort_values(ascending=False)

missing_percentage
```

## **Outlier**

```python
numeric_columns = [
    'number_of_reviews',
    'current/discounted_price'
]

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print("=" * 60)
    print(f"Column: {col}")
    print(f"Q1: {Q1:.2f}")
    print(f"Q3: {Q3:.2f}")
    print(f"IQR: {IQR:.2f}")
    print(f"Lower Bound: {lower_bound:.2f}")
    print(f"Upper Bound: {upper_bound:.2f}")
    print(f"Number of Outliers: {len(outliers)}")
```

```python
# Convert listed_price temporarily to numeric
listed_price_num = pd.to_numeric(
    df['listed_price']
    .astype(str)
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False),
    errors='coerce'
)

# Calculate Discount Percentage
df['discount_percentage'] = (
    (listed_price_num - df['current/discounted_price'])
    / listed_price_num
) * 100
```

#Statistical Test

```python
low_discount = df[df['discount_percentage'] < 20]['number_of_reviews'].dropna()

high_discount = df[df['discount_percentage'] >= 20]['number_of_reviews'].dropna()

print("Low Discount:", len(low_discount))
print("High Discount:", len(high_discount))
```

```python
print("Average Reviews - Low Discount:", low_discount.mean())
print("Average Reviews - High Discount:", high_discount.mean())
```

```python
from scipy.stats import ttest_ind

t_stat, p_value = ttest_ind(
    low_discount,
    high_discount,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

# The difference in average customer engagement between low-discount and high-discount products was not statistically significant (p = 0.369). Therefore, there is insufficient evidence to conclude that discount level is associated with a significant difference in the number of customer reviews
